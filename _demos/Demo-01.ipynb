The idea is that we can add precise descriptions of code - beyond the code comments, to document what code is intended to do, why we are doing it, and any assumptions/choices made in implementing it.

### Scope
As you may have noticed above, the scope of variables in a cell are global. They can be accessed across the notebook.

**Gotcha Warning:**  This means that variables can be accidentally overwritten if we are not careful! Use thoughtful variable names to avoid this.

````python
y = 5
````

````python
# Overwrite the current value of x
x = 10
````

Similarly, the scope of functions is global across the notebook. However, the variables inside our functions are local.

````python
# Define a simple function.

def my_func(x, y):
    if x > y:
        return x
    else:
        return y
````

````python
my_func(1,2)
````

````python
# We can re-define functions, overwriting previous versions
def my_func(x, y):
    return (x-1, y+2)
````

````python
# What is in scope here?
(a, b) = my_func(1, 2)
````

````python
print(a)
print(b)
````

Python often uses a "lazy evaluation" strategy where it only evaluates an expression if it is actually used. For example, `enumerate(my_arr)` will iterate over all elements in a list. However, notice below, that it initiall returns an "enumerate object". This just means that it will not actually evaluate the enumeration until we explicilty need it to, as in the following cell.

````python
# enumerate: returns iterable tuple (index, element) of a list
a = enumerate( ['311', '320', '330'] )
print(a)
````

````python
# Recall here that Python3 does lazy evaluation for these iterators.  We have to manually expand it.
list(a)
````

We can iterate over the enumerate object using for loops.

````python
a = ['311', '320', '330']
for i,j in enumerate(a):
    print(i,j)
````

When transforming data for analysis, we may want to apply the same function across a set of values. The `map` function is helpful for this.

````python
def squared(x):
    return x**2
````

````python
## Map and Filter

# map: apply a function to a sequence or iterable

arr = [1, 2, 3, 4, 5]
out = map(squared, arr)
print(list(out))
print(arr)
````

List comprehensions are a concise way of creating lists by appling an `expression` to each `item` in an `iterable` (e.g. a list). They may also include an optional `if condition` to only return select values from the iterable. They are expressed in the form `[expression for item in iterable (optional) if condition]`.

In the cell below, we are using list comprehension to cube the numbers in `arr` that are less than or equal to three. In this example,
- `expression` corresponds to `x**3`
- `item` corresponds to `x`
- `iterable` corresponds to `arr`
- `if condition` corresponds to `if x <=3`

````python
# create a new array using list comprehension
new_arr = [x**3 for x in arr if x <= 3]
````

````python
new_arr
````

Lambda functions are small anonymous functions that do not have a defined name. They have hte form `lambda x : expr(x)`.
We will often use lambda functions to apply an expression across a subset of our data, e.g., to transform all values in a column.

Below, we use a define a lambda function to square our values rather than writing the explicit function above.

````python
# What happened here??
arr = [1, 2, 3, 4, 5]
out = map(lambda x: x**2, arr)
print(out)


# Remember again lazy evaluation!
print(list(out))
````

The filter function takes in a function, `f`, that tests if a variable, `x`, meets a condition (returning true or false), and a list of values to test. It returns only those where `f(x)` returns `True`.

````python
# filter: returns a list of elements for which a predicate is true

arr = [1, 2, 3, 4, 5, 6, 7]
out = filter(lambda x: x % 2 == 0, arr)
print(out)

# Remember again that we have to explicitly evaluate the iterator.
print(list(out))
````

````python
# A more pythonic way: list comprehension...
x = [i for i in arr if i % 2 == 0]
x
````
